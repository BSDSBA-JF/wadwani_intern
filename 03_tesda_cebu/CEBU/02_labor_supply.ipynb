# CEBU - Labor Supply

Based on the *"October 24 - Questions for TESDA"* gdoc, this should be the TESDA EGAC. Some other notes:
- we are currently in year 3
- TESDA Cebu's analysis prioritizes IT-BPO and Manufacturing as the key sectors to feature

Code below gets the totals (E, G, A, C) with the following filters:
- sector "INFORMATION AND COMMUNICATION TECHNOLOGY" with subsector "CONTACT CENTER AND BPO"
- sector "MANUFACTURING"
- Considering it's just TESDA, I only used the EGAC values from the "Institution Based" section

## Total Enrolled (E)

```python
# Load the CSV
file_path = "data/[Provincial] Reduction - Cebu - Section IV (R).csv"
df = pd.read_csv(file_path, header=4)

# Standardize text for text under these headers
df["Sector"] = df["Sector"].astype(str).str.strip().str.upper()
df["Subsector"] = df["Subsector"].astype(str).str.strip().str.upper()

# Column for Year 3 -> Institution Based -> E
target_column = "495,393"

# To make comma numbers numeric instead (1,713 for example)
def clean_numeric(series):
    return (
        series.astype(str)
        .str.replace(",", "", regex=False)
        .str.strip()
        .replace("", pd.NA)
        .astype(float)
    )

# ICT-BPO
ict_bpo = df[
    (df["Sector"] == "INFORMATION AND COMMUNICATION TECHNOLOGY") &
    (df["Subsector"] == "CONTACT CENTER AND BPO")
].copy()

ict_bpo[target_column] = clean_numeric(ict_bpo[target_column])
ict_total = ict_bpo[target_column].sum()

# MANUFACTURING (all subsectors)
mfg = df[df["Sector"] == "MANUFACTURING"].copy()
mfg[target_column] = clean_numeric(mfg[target_column])
mfg_total = mfg[target_column].sum()

# Grand total 
grand_total = ict_total + mfg_total

print(f"ICT - Contact Center & BPO (E): {ict_total:,.0f}")
print(f"MANUFACTURING (E): {mfg_total:,.0f}")
print(f"Grand Total (E): {grand_total:,.0f}")
```
