# CEBU - Labor Supply

loading the file

```python
file_path = "data/EG 2025.csv"
df = pd.read_csv(file_path)
```

entries under column "Region Name"

```python
# Get unique region names
unique_regions = df["Region Name"].dropna().unique()

# Print or save them
print(unique_regions)
```

## Total Enrolled (E)

```python
# Filter rows for Region VII - Central Visayas
region_vii = df[df["Region Name"] == "Region VII - Central Visayas"]

# Calculate total enrolled
total_enrolled = region_vii["Enrolled"].sum()

print("Total Enrolled (Region VII - Central Visayas):", total_enrolled)
```

## Total Graduated (G)

```python
# Filter rows for Region VII - Central Visayas
region_vii = df[df["Region Name"] == "Region VII - Central Visayas"]

# Calculate total graduates
total_enrolled = region_vii["Graduates"].sum()

print("Total Graduates (Region VII - Central Visayas):", total_enrolled)
```
