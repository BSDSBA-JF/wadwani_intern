For the Unemployment rate:

Everyone not unemployed, is employed, but now we have to determine what kind of employed:
For the Self-Employed, look into this column:
* PUFC23_PCLASS	C23-Class of Worker (Primary Occupation) for January OR AUGUST

| Class of Worker                     | Code |
| ----------------------------------- | ---- |
| Private Household                   | 0    |
| Private Establishment               | 1    |
| Gov't / Gov't Corporation           | 2    |
| Self Employed                       | 3    |
| Employer                            | 4    |
| With pay (Family owned Business)    | 5    |
| Without Pay (Family owned Business) | 6    |

For the Underemployed, it includes both time-related underemployment (insufficient hours), skill/quality mismatch. However, the PSA mainly measures the time-related aspect. The relevant column is:
* PUFC21_PLADDW	C21-Look for Additional Work (1 - Yes, 2 - No)

For the TESDA or not, we use:
PUFC09_GRADTECH	C09-Graduate of technical/vocational course for January/August

For the pathway, we will determine that through the below:
* PUFC07_GRADE	C07 - Highest Grade Completed for January/August
* PUFC07_GRADE_VS2	- Highest Grade Completed for February/March

| Education Level    | Status        | Code(s)                                         |

```python
relevant_columns = [
    "PUFC05_AGE",
    "PUFC11_WORK", 
    "PUFC23_PCLASS", 
    "PUFC21_PLADDW", 
    "PUFC09_GRADTECH",
    "PUFC07_GRADE",
    "PUFC08_CURSCH",
]

formal_columns = [
    "Age",
    "Work Indicator",
    "Class of Worker",
    "Is Underemployed",
    "Is TESDA",
    "Grade",
    "Is Study"
]
```

Because we need to know if they wanted additional work (for underemployed) and class of work (for self-employed), then we need to drop the nulls. From the original 345_341 data points, we are left with 153_736 (44.51%)

```python
# Cleaning time
combined_cleaned_df = labor_force_df[relevant_columns].dropna()
combined_cleaned_df.columns = formal_columns
combined_cleaned_df.loc[:, 'Grade'] = combined_cleaned_df['Grade'].astype(int)
```

Coding out actual ranges

```python
# Education code ranges
education_ranges = {
    "No grade completed": [(0, 1000), (2000, 2000)],
    "Elementary Undergraduate": [(10011, 10015), (10004, 10006)],
    "Elementary Graduate": [(10018, 10018)],
    "Junior High School Undergraduate": [(24011, 24013), (24004, 24006)],
    "Junior High School Graduate": [(24014, 24015)],
    "Senior High School Undergraduate": [(34011, 34011), (34021, 34021), (34031, 34031), (35011, 35011)],
    "Senior High School Graduate": [(34013, 34013), (34023, 34023), (34033, 34033), (35013, 35013)],
    "Post Secondary Undergraduate": [(40001, 40003), (50001, 50003)],
    "Post Secondary Graduate": [(40011, 49999), (50011, 59999)],
    "College Undergraduate": [(60001, 60006)],
    "College Graduate": [(60000, 60000), (60011, 69999), (70000, 79999), (80000, 89999)]
}

def classify_education(code):
    """Classify the code to what grade it was completed"""
    for level, ranges in education_ranges.items():
        for lower, upper in ranges:
            if lower <= code <= upper:
                return level
    return "Unknown"

# Mapping to simplified education levels: Undergraduate vs Graduate
grade_map = {
    "Junior High School Graduate": "Junior High School",
    "College Graduate": "Higher Education Institution",
    "Elementary Graduate":"Elementary",
    "Elementary Undergraduate":"No grade completed",
    "Junior High School Undergraduate":"Elementary",
    "College Undergraduate":"Senior High School",
    "Post Secondary Graduate":"Higher Education Institution",
    "Senior High School Graduate":"Senior High School",
    "No grade completed": "No grade completed",
    "Senior High School Undergraduate": "Junior High School",
    "Post Secondary Undergraduate":"Senior High School"

}

# Apply to your DataFrame
combined_cleaned_df['Grade Decoded'] = combined_cleaned_df['Grade'].map(classify_education)
combined_cleaned_df['Grade Aggregated'] = combined_cleaned_df['Grade Decoded'].map(grade_map)


code_mapping = {"0":"Private Household", 
           "1":"Private Establishment", 
           "2":"Gov't / Gov't Corporation", 
           "3":"Self Employed", 
           "4":"Employer", 
           "5":"With pay (Family owned Business)", 
           "6":"Without Pay (Family owned Business)"
}
combined_cleaned_df["Class of Worker Decoded"] = combined_cleaned_df["Class of Worker"].map(code_mapping)
```
