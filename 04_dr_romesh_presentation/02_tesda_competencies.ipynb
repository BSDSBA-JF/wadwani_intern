```python
# open file
filepath = 'data/QSO Competency Bank(beta)_Competencies - TR_Table.csv'
competencies_df = pd.read_csv(filepath).loc[1:]

# remove superseded rows
is_current = ~competencies_df['Qualification Title'].str.contains("Superseded")
competencies_df = competencies_df[is_current]

# Remove rows without qualification code
has_quali_code = ~(competencies_df["Qualification Code"] == '-')
competencies_df = competencies_df[has_quali_code]

# Get for competencies three and four
is_NC_III_IV = competencies_df["Qualification Title"].str.contains(r"NC III|NC IV", na=False)
competencies_III_IV_df = competencies_df[is_NC_III_IV]
competencies_I_II_df = competencies_df[~is_NC_III_IV]
```
