```python
import pandas as pd
import numpy as np
```

```python
# Open the datasets
filepath = 'data/PHL-PSA-LFS-2025-01-PUF/LFS PUF January 2025.CSV'
jan_df = pd.read_csv(filepath)
jan_df.replace(' ', np.nan, inplace=True)

filepath = 'data/PHL-PSA-LFS-2025-04-PUF/LFS PUF April 2025.CSV'
apr_df = pd.read_csv(filepath)
apr_df.replace(' ', np.nan, inplace=True)

combined_df = pd.concat([jan_df, apr_df], axis=0)

combined_df.loc[:, 'PUFC43_QKB'] = combined_df['PUFC43_QKB'].replace(r'^\s*$', np.nan, regex=True)
```
