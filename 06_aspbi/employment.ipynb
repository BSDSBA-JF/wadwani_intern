```python
import pandas as pd
import os
```

```python
excel_files = [f for f in sorted(os.listdir('employment'))
               if not f.startswith("~$")] # to remove the temporary files
 
results = []
for file in excel_files:
    excel = pd.ExcelFile('employment/' + file)
    results.append(len(excel.sheet_names))
```
