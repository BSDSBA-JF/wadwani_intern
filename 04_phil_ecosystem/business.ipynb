```python
is_lf = (apr_df['PUFNEWEMPSTAT'] == '1') | (apr_df['PUFNEWEMPSTAT'] == '2')
lf_df = apr_df[is_lf].copy()
lf_df['PUFC16_PKB'] = lf_df['PUFC16_PKB'].replace('  ', np.nan)
```

```python
business_ranges = {
    "A": [(1, 3)],
    "B": [(5, 9)],
    "C": [(10, 33)],
    "D": [(35, 35)],
    "E": [(36, 39)],
    "F": [(41, 43)],
    "G": [(45, 47)],
    "H": [(49, 53)],
    "I": [(55, 56)],
    "J": [(58, 63)],
    "K": [(64, 66)],
    "L": [(68, 68)],
    "M": [(69, 75)],
    "N": [(77, 82)],
    "P": [(85, 85)],
    "Q": [(86, 88)],
    "R": [(90, 93)],
    "S": [(94, 96)],
    "Other": [(84, 84), (97, 98), (99, 99)]
}

def classify_business(code):
    if pd.isna(code):
        return "Other"

    code = int(code)
    for letter, ranges in business_ranges.items():
        for lower, upper in ranges:
            if lower <= code <= upper:
                return letter
    return "Other"

lf_df['Business_Letter'] = (
    lf_df['PUFC16_PKB']
    .map(classify_business)
)
```

```python
is_private = lf_df['PUFC23_PCLASS'] == '1'
private_df = lf_df[is_private]
```
